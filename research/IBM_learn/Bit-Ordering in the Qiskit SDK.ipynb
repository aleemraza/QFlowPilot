Lesson: Bit-Ordering in the Qiskit SDK
This is an important Qiskit concept. Before working with QAOA, quantum circuits, measurement results, and quality estimation, you must understand which bit represents which qubit.

The main difficulty is that Qiskit uses different visual arrangements for circuit diagrams and bit strings. The qubit labels do not change, but their displayed positions can be confusing.

Let's understand it step by step.

For a three-qubit circuit:
Qubit	Position in diagram	Position in bit string	Integer value
q0	Top	Rightmost	\(2^0=1\)
q1	Middle	Middle	\(2^1=2\)
q2	Bottom	Leftmost	\(2^2=4\)

For example, the bit string 101 means:
- q2 = 1
- q1 = 0
- q0 = 1
Its decimal value is:
\[
1\times 2^2+0\times 2^1+1\times 2^0=5
\]

Remember: Qiskit writes the most significant bit on the left and the least significant bit on the right.

Qubit labels in QuantumCircuit
Qiskit stores qubits in a list:

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(2)

print(qc.qubits)
print(qc.qubits[0])
print(qc.qubits[1])
```

Circuit diagrams: top-to-bottom ordering
Consider this circuit:

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(2)

qc.x(1)

print(qc.draw())
```

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler

qc = QuantumCircuit(2)

qc.x(1)
qc.measure_all()

sampler = StatevectorSampler()
job = sampler.run([qc])
result = job.result()

print(result[0].data.meas.get_counts())
```
